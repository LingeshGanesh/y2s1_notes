# **Business Process Analysis**

# **_Static Analysis_**


Static Analysis in Business Process Management requires analyzing the issue of business processes without process execution. This usually involves (a) reviewing the interview scripts and (b) reviewing the as-is models.

## **Static Reports**

Static reports help us to:
- Identify the costs and time about the process
- Identify the issues with respect to activities and resources

Examples of static reports include
1) Cycle Time Calculation
2) Cost Calculation
3) Resource Calculation
4) Process Data Tables => roles and activities
5) Root Cause Analysis ( [RCI](rci.ipynb) )

# **_Dynamic Analysis_**

Dynamic analysis of business processes are done by executing the business process. This usually involves (a) pilot testing and (b) running simulations based on the process model.

**Why Dynamic instead of just Static**  
Static analysis may not offer the metrics needed to evaluate the performance goal, and certain analysis reports may not be producible through static analysis alone.

Dynamic analysis can determine the best solution or outcome among various choices and utilize quantitative data to enable the analysis and comparison of potential value across process variations.
