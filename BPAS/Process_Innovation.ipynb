# Digital Twins
A Digital Twin is a dynamic digital representation of a physical asset, process or system, updates using real-world data, often through IoT sensors. They are linked to their real-world counterparts and are used to understand the state of the thing of system, respond to changes, improve operations and add value.

# Internet of Things (IoT)
IoT refers to connected physical devices that sense or act on their environment and exchange data with digital systems.

IoT devices are given a land and task in BPMN diagrams.
