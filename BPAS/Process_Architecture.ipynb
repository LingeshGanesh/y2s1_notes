# **Process Architecture**


## Functional Units

Functional units help operate the process. Multiple functional units from various departments can be used to administer these processes. 

Process ownership is important to identify 
1) Responsibility for performing the process or task
2) Accountability/approval authority for the results achieved by the process or task

## **Process Hierarchy Model**

A Process Hierarchy model is used to show the hierarchy of sub-processes in the Executive Model that are related to a process along with descriptions. This is because Process Executive Models usually only depict until level 3 to manage diagram complexity, and Process Hierarchy models are used for higher levels beyond that.
