# **Value Chain**

The value of a product refers to a customer's willingness to pay. The processes that generate value in any organization is the value chain.

Organizations typically support from 3 to 15 value chains as it helps to:
 - Identify sources of competitive advantage
 - Systemetically examine the processes in an organization, their interactions and impact on business performance by each other
 - Align investments with the value chain
