# **Solution Blueprints**

The objective of a solution blue print is to communicate the system 'vision' **without committing to very specific implementation details** over which we should have control.

- Contains the specifications of various aspects of an IT solution
- Each model emphasizes aspects that are meaningful to certain stakholders
- Initial, high-level view of complete end-to-end solution
- Not concerned with constraints/technical implementations or detailed designs

# Concept Solution Blueprints

Concept Solution Blue prints is a combination of various elements that describe systems' interfaces and design, application's interactions, information management, cost management and risk management.

They are made up of 3 main components:
1) Architecture Model
2) Risk Model
3) Cost Model

## **Architecture Model**

Architecture models are used to describe the overall picture of the solution in terms that all stakeholders can easily understand. It defines the scope of the solution, depicting actors, systems, interfaces and the major functions of the proposed solution.

## **Risk Models**

Risk is defined as "an uncertain event or condition that, if it occurs, has a positive or negative impact on a project objective". The Concept Risk Model is used to provide a guideline for identifying, quantifying risks and deriving mitigation strategies.

# **Cost Modelling**

Cost models are computational approaches to estimate the effort of production of (software) products, financially. It considers detail on the specifics of the cost of process modelling and analysis and aids understanding of potential qualitiative cost level for designing, implementing and maintaining the model. 

Concept cost models provide an estimated qualitative cost level for the proposed solution.
