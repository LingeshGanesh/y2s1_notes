# **Dynamic Analysis**

## **Establish Process Performance Goals**
We first begin by identifying the important factors to address in this process such as cost, effciency, quality, etc. This involves things like determining which factors are the most important and what are the measurable goals for the organization. It is usually done in the following steps:

1) Identify the process performance goals and their properties
2) Identify any binding organization policies or management decisions related to the process
3) Document the process performance goals and measurement metrics for each of the goals

## **Modelling Activities and Simulations**

In this phase, we identify the current problems (bottlenecks) as well as the critical areas to focus on. This is also the phase in which we explore the various alternatives for the To-Be process.

#### **Prepare Simulations**
4 types of data are needed for simulation purposes:
1) **Process-wide:** Triggers, start times, arrival patterns, rate, termination conditions, dependecies
2) **Activity-related:** Resource requirements, required roles/individuals, time (constant, exponential, etc.), cost incurred, queue handling
3) **Resource-related:** Resource costs and quantity, work hours, maximum utilization
4) **Decision points:** What are the probability of decision points based on past history or assumptions

## **Impact Analysis of To-Be Process Alternatives**
In this phase we will identify who will be affected by the change in the process. The impact of the changes introduced by each of the To-Be process alternatives are analyzed by the following:

- Internal impact
- External impact
- Hard quantifiable benefits/losses
- Soft qualitative benefits/losses

These will influence the final design choice of the process. The best to-be alternative may prove to be infeasible to implement since it has major internal and external impacts.

Greater Impact => Higher Risk => Potentially Higher Implementation Cost/Time => Resistance to Change

## **Final Selection of To-Be Process**
The final decision regarding the most suitable process that should be chosen as the To-Be process for implementation should be based on the following criterias:

- Performance Goals
- Simulation Results
- Impact Analysis
- Management Decisions
