# **Process Innovation**


## **Types of Business Model Changes**

**Incremental or Continuous Improvements**
- Generate limited value
- May involve some risks
- May involve some level of cultural and structural change
- Impacts a limited portion of an organization

**Radical Innovation**
- increase value by 50%, 100% or more (game changing)
- Riskier than incremental improvements
- Requires a higher level of cultural and structural change
- Impacts a broader portion of an organization
