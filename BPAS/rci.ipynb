# **Root Cause Analysis (RCI) Model**

1) Identify the relevant issue for the process and give a short description
2) Identify the cause of the issue
3) Identify the common cause of each causes of the root cause

## **Issue Categories**

**Process-related:** Involves current processes in the workflow, such as: non-automation, missing activities, missing participats, wrong activity sequence, etc.

**System-related:** Improper performance, interface, functionality, etc. relating to IT applications

**People-related:** Work attitude of employees, lack of knowledge/skills/training, etc.

**Policy-related:** Relating to laws or policies that affect the business process. These can be 
internal (organization or departmental rules) or external (government or other regulatory bodies)

**Others:** External partners or those that cannot be placed in the above categories
